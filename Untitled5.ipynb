```python
import numpy as np

# a. Basic ndarray
basic_array = np.array([1, 2, 3, 4, 5])
print("Basic ndarray:\n", basic_array)

# b. Array of zeros
zeros_array = np.zeros((3, 3))
print("\nArray of zeros:\n", zeros_array)

# c. Array of ones
ones_array = np.ones((2, 2))
print("\nArray of ones:\n", ones_array)

# d. Random numbers in ndarray
random_array = np.random.rand(2, 2)
print("\nRandom numbers in ndarray:\n", random_array)

random_integers = np.random.randint(1, 10, (3, 3))
print("\nRandom integers in ndarray:\n", random_integers)

# e. An array of your choice
my_array = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print("\nMy custom array:\n", my_array)

# f. Identity matrix in NumPy (Imatrix)
identity_matrix = np.eye(3)
print("\nIdentity matrix:\n", identity_matrix)

# g. Evenly spaced ndarray
evenly_spaced_array1 = np.arange(0, 10, 2)
print("\nEvenly spaced array (arange):\n", evenly_spaced_array1)

evenly_spaced_array2 = np.linspace(0, 1, 5)
print("\nEvenly spaced array (linspace):\n", evenly_spaced_array2)
```

```python
arr=np.array([[1,2,3,4,5,6]])
print("a,number of dimensions:",arr.ndim)
print("b,shape of array:",arr.shape)
print("c,size of array:",arr.size)

reshaped_arr=arr.reshape(2,3)
print("d,reshaped array(3*2):\n",reshaped_arr)
```

```python
import numpy as np

arr=np.array([1,2,3])

col_vector=np.expand_dims(arr,axis=1)
row_vector=np.expand_dims(arr,axis=0)

arr_expanded=np.array([[[1],[2],[3]]])
squeezed_arr=np.squeeze(arr_expanded)

arr_2d=np.array([[3,1,2],[6,4,5]])
sorted_row=np.sort(arr_2d,axis=1)
sorted_col=np.sort(arr_2d,axis=0)
indices=np.argsort(arr_2d,axis=1)

print("column vector:\n",col_vector)
print("row vector:\n",row_vector)
print("squeezed array:\n",squeezed_arr)
print("sorted array(row-wise):\n",sorted_row)
print("sorted array(col-wise):\n",sorted_col)
print("sorting indices:\n",indices)
```
